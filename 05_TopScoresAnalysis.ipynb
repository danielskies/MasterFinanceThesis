```python
import pandas as pd
import numpy as np
import warnings
warnings.filterwarnings('ignore')

# File path
file_path = "/Users/daniel/Library/Mobile Documents/com~apple~CloudDocs/Documents/Master Finance/MasterThesis/ThesisData/perfect_balanced_ai_dataset.csv"

def convert_letter_grades_to_numeric(grade):
    """
    Convert letter grades to numeric scores
    A = 5, B = 4, C = 3, D = 2, E = 1
    """
    if pd.isna(grade) or grade == '':
        return np.nan
    
    grade_mapping = {
        'A': 5, 'B': 4, 'C': 3, 'D': 2, 'E': 1
    }
    
    return grade_mapping.get(str(grade).upper().strip(), np.nan)

def print_company_details(company_data, rank, sector=None):
    """
    Print detailed company information including all model summaries and terms
    """
    ticker = company_data['Ticker']
    company_name = company_data['Company Name']
    year = company_data['Year']
    filing_date = company_data['filingDate']
    cumulative_score = company_data.get('cumulative_ai_score_overall', 'N/A')
    average_score = company_data.get('average_ai_score_overall', 'N/A')
    
    # Header
    if sector:
        print(f"#{rank}. {ticker} - {company_name}")
    else:
        print(f"🏆 #{rank}. {ticker} - {company_name}")
    
    print(f"     📅 Year: {year} | Filing Date: {filing_date}")
    print(f"     🎯 Cumulative AI Score: {cumulative_score:.2f} | Average: {average_score:.2f}")
    print(f"     📊 Sector: {company_data['Sector']}")
    
    # Show individual model scores
    models = ['flash1_5', 'flash2_5', 'gpt4o']
    model_names = ['Gemini 1.5 Flash', 'Gemini 2.5 Flash', 'GPT-4o']
    
    model_scores = []
    for model, name in zip(models, model_names):
        score = company_data.get(f'cumulative_ai_score_{model}', 'N/A')
        if score != 'N/A':
            model_scores.append(f"{name}: {score:.1f}")
        else:
            model_scores.append(f"{name}: N/A")
    
    print(f"     💻 Model Scores: {' | '.join(model_scores)}")
    print()
    
    # Show detailed AI factor scores for each model
    ai_factors = [
        'AI Washing Index',
        'Disclosure Sentiment', 
        'Forward-Looking',
        'Risk - External Threats',
        'Risk - Non-Adoption', 
        'Risk - Own Adoption',
        'Strategic Depth',
        'Talent & Investment'
    ]
    
    print(f"     📊 DETAILED AI FACTOR SCORES:")
    print(f"     {'-' * 70}")
    
    for model, model_name in zip(models, model_names):
        print(f"     🤖 {model_name}:")
        
        factor_scores = []
        for factor in ai_factors:
            # Get original categorical score
            categorical_col = f"{factor}_{model}"
            # Get numeric score
            numeric_col = f"{factor}_{model}_numeric"
            
            if categorical_col in company_data.index:
                categorical_score = company_data[categorical_col]
                numeric_score = company_data.get(numeric_col, 'N/A')
                
                if pd.notna(categorical_score) and numeric_score != 'N/A':
                    factor_scores.append(f"{factor}: {categorical_score}({numeric_score:.0f})")
                elif pd.notna(categorical_score):
                    factor_scores.append(f"{factor}: {categorical_score}")
                else:
                    factor_scores.append(f"{factor}: N/A")
            else:
                factor_scores.append(f"{factor}: N/A")
        
        # Print factor scores in a formatted way (2 per line for readability)
        for i in range(0, len(factor_scores), 2):
            line_scores = factor_scores[i:i+2]
            if len(line_scores) == 2:
                print(f"        {line_scores[0]:<35} | {line_scores[1]}")
            else:
                print(f"        {line_scores[0]}")
        print()
    
    print()
    
    # Full summaries for all models
    print(f"     📄 FULL AI SUMMARIES:")
    print(f"     {'-' * 50}")
    
    for model, name in zip(models, model_names):
        summary_col = f'Overall Summary_{model}'
        if summary_col in company_data.index and pd.notna(company_data[summary_col]):
            summary = str(company_data[summary_col])
            print(f"     🤖 {name}:")
            # Wrap text nicely
            words = summary.split()
            lines = []
            current_line = []
            current_length = 0
            
            for word in words:
                if current_length + len(word) + 1 <= 100:  # 100 chars per line
                    current_line.append(word)
                    current_length += len(word) + 1
                else:
                    if current_line:
                        lines.append(' '.join(current_line))
                    current_line = [word]
                    current_length = len(word)
            
            if current_line:
                lines.append(' '.join(current_line))
            
            for line in lines:
                print(f"        {line}")
            print()
        else:
            print(f"     🤖 {name}: No summary available")
            print()
    
    # Full key terms for all models
    print(f"     🔑 KEY AI TERMS:")
    print(f"     {'-' * 50}")
    
    for model, name in zip(models, model_names):
        terms_col = f'Key AI Terms_{model}'
        if terms_col in company_data.index and pd.notna(company_data[terms_col]) and str(company_data[terms_col]).strip():
            terms = str(company_data[terms_col])
            print(f"     🏷️ {name}: {terms}")
        else:
            print(f"     🏷️ {name}: No specific AI terms identified")
    
    print(f"\n     {'=' * 80}\n")

def enhanced_ai_analysis():
    """
    Enhanced AI analysis with full summaries and terms
    """
    print("🚀 ENHANCED AI SCORE ANALYSIS - FULL DETAILS")
    print("=" * 100)
    
    # Load the dataset
    try:
        df = pd.read_csv(file_path)
        print(f"✅ Loaded dataset: {df.shape[0]:,} rows × {df.shape[1]:,} columns")
    except Exception as e:
        print(f"❌ Error loading dataset: {e}")
        return None
    
    # Define AI factor columns for each model
    ai_factors = [
        'AI Washing Index',
        'Disclosure Sentiment', 
        'Forward-Looking',
        'Risk - External Threats',
        'Risk - Non-Adoption', 
        'Risk - Own Adoption',
        'Strategic Depth',
        'Talent & Investment'
    ]
    
    models = ['flash1_5', 'flash2_5', 'gpt4o']
    
    print(f"\n📊 CONVERTING LETTER GRADES TO NUMERIC SCORES")
    print("-" * 60)
    print("Grade mapping: A=5, B=4, C=3, D=2, E=1")
    
    # Convert all AI factor scores to numeric
    for factor in ai_factors:
        for model in models:
            col_name = f"{factor}_{model}"
            if col_name in df.columns:
                # Create numeric version
                numeric_col_name = f"{factor}_{model}_numeric"
                df[numeric_col_name] = df[col_name].apply(convert_letter_grades_to_numeric)
    
    print(f"\n🧮 CALCULATING CUMULATIVE AI SCORES")
    print("-" * 60)
    
    # Calculate cumulative AI scores for each model
    for model in models:
        numeric_cols = [f"{factor}_{model}_numeric" for factor in ai_factors 
                       if f"{factor}_{model}_numeric" in df.columns]
        
        if numeric_cols:
            # Calculate cumulative score (sum of all factors)
            df[f'cumulative_ai_score_{model}'] = df[numeric_cols].sum(axis=1, skipna=True)
            
            # Calculate average score (mean of all factors)
            df[f'average_ai_score_{model}'] = df[numeric_cols].mean(axis=1, skipna=True)
            
            print(f"   {model}: {len(numeric_cols)} factors → Cumulative range: "
                  f"{df[f'cumulative_ai_score_{model}'].min():.1f}-{df[f'cumulative_ai_score_{model}'].max():.1f}")
    
    # Calculate overall cumulative and average scores across all models
    cumulative_cols = [f'cumulative_ai_score_{model}' for model in models]
    average_cols = [f'average_ai_score_{model}' for model in models]
    
    if all(col in df.columns for col in cumulative_cols):
        df['cumulative_ai_score_overall'] = df[cumulative_cols].mean(axis=1, skipna=True)
        print(f"\n🎯 Overall Cumulative AI Score: {df['cumulative_ai_score_overall'].min():.2f}-{df['cumulative_ai_score_overall'].max():.2f}")
    
    if all(col in df.columns for col in average_cols):
        df['average_ai_score_overall'] = df[average_cols].mean(axis=1, skipna=True)
        print(f"🎯 Overall Average AI Score: {df['average_ai_score_overall'].min():.2f}-{df['average_ai_score_overall'].max():.2f}")
    
    # OVERALL TOP 5 COMPANIES
    print(f"\n🌟 OVERALL TOP 5 AI COMPANIES (ALL SECTORS)")
    print("=" * 100)
    
    if 'cumulative_ai_score_overall' in df.columns:
        top_5_overall = df.nlargest(5, 'cumulative_ai_score_overall')
        
        for rank, (idx, company) in enumerate(top_5_overall.iterrows(), 1):
            print_company_details(company, rank)
    
    # TOP 3 COMPANIES BY SECTOR
    print(f"\n🏆 TOP 3 AI COMPANIES BY SECTOR (DETAILED ANALYSIS)")
    print("=" * 100)
    
    sectors = df['Sector'].unique()
    print(f"Analyzing {len(sectors)} sectors...")
    
    for sector in sorted(sectors):
        sector_data = df[df['Sector'] == sector].copy()
        
        if 'cumulative_ai_score_overall' in sector_data.columns:
            # Sort by overall cumulative AI score (descending)
            top_3 = sector_data.nlargest(3, 'cumulative_ai_score_overall')
            
            print(f"\n🏢 SECTOR: {sector.upper()}")
            print("=" * 80)
            
            for rank, (idx, company) in enumerate(top_3.iterrows(), 1):
                print_company_details(company, rank, sector)
    
    # Score distribution analysis
    print(f"\n📈 STATISTICAL ANALYSIS")
    print("=" * 60)
    
    if 'cumulative_ai_score_overall' in df.columns:
        score_stats = df['cumulative_ai_score_overall'].describe()
        print("Overall Cumulative AI Score Statistics:")
        print(f"   Mean: {score_stats['mean']:.2f}")
        print(f"   Median: {score_stats['50%']:.2f}")
        print(f"   Std Dev: {score_stats['std']:.2f}")
        print(f"   Range: {score_stats['min']:.2f} - {score_stats['max']:.2f}")
        
        # Percentiles
        percentiles = df['cumulative_ai_score_overall'].quantile([0.1, 0.25, 0.75, 0.9])
        print(f"   10th percentile: {percentiles[0.1]:.2f}")
        print(f"   25th percentile: {percentiles[0.25]:.2f}")
        print(f"   75th percentile: {percentiles[0.75]:.2f}")
        print(f"   90th percentile: {percentiles[0.9]:.2f}")
    
    # Model correlation analysis
    print(f"\n🔄 MODEL CORRELATION ANALYSIS")
    print("-" * 40)
    
    model_score_cols = [f'cumulative_ai_score_{model}' for model in models if f'cumulative_ai_score_{model}' in df.columns]
    if len(model_score_cols) > 1:
        correlation_matrix = df[model_score_cols].corr()
        print("Correlation between model scores:")
        print(correlation_matrix.round(3))
    
    # Sector analysis
    print(f"\n🏭 SECTOR ANALYSIS")
    print("-" * 40)
    
    if 'cumulative_ai_score_overall' in df.columns:
        sector_stats = df.groupby('Sector')['cumulative_ai_score_overall'].agg(['mean', 'std', 'count']).round(2)
        sector_stats = sector_stats.sort_values('mean', ascending=False)
        print("Average AI scores by sector:")
        print(sector_stats.to_string())
    
    print(f"\n✅ ENHANCED ANALYSIS COMPLETE")
    print("=" * 100)
    print("💡 Key Insights:")
    print("   • Full summaries show how each model interprets AI strategies")
    print("   • Key terms reveal specific AI technologies mentioned")
    print("   • Top companies demonstrate diverse AI approaches across sectors")
    print("   • Model correlations indicate consistency in AI assessment")
    
    return df

# Execute the enhanced analysis
if __name__ == "__main__":
    enhanced_df = enhanced_ai_analysis()
```

```python
import pandas as pd
import numpy as np
import warnings
warnings.filterwarnings('ignore')

# File path from your original script
file_path = "/Users/daniel/Library/Mobile Documents/com~apple~CloudDocs/Documents/Master Finance/MasterThesis/ThesisData/perfect_balanced_ai_dataset.csv"

def convert_letter_grades_to_numeric(grade):
    """
    Convert letter grades to numeric scores
    A = 5, B = 4, C = 3, D = 2, E = 1
    """
    if pd.isna(grade) or grade == '':
        return np.nan
    
    grade_mapping = {
        'A': 5, 'B': 4, 'C': 3, 'D': 2, 'E': 1
    }
    
    return grade_mapping.get(str(grade).upper().strip(), np.nan)

def print_company_details(company_data, rank, sector=None):
    """
    Print detailed company information including all model summaries and terms
    """
    ticker = company_data['Ticker']
    company_name = company_data['Company Name']
    year = company_data['Year']
    filing_date = company_data['filingDate']
    cumulative_score = company_data.get('cumulative_ai_score_overall', 'N/A')
    average_score = company_data.get('average_ai_score_overall', 'N/A')
    
    # Header
    print(f"📄 #{rank}. {ticker} - {company_name}")
    
    print(f"     📅 Year: {year} | Filing Date: {filing_date}")
    print(f"     🎯 Cumulative AI Score: {cumulative_score:.2f} | Average: {average_score:.2f}")
    
    # Show individual model scores
    models = ['flash1_5', 'flash2_5', 'gpt4o']
    model_names = ['Gemini 1.5 Flash', 'Gemini 2.5 Flash', 'GPT-4o']
    
    model_scores = []
    for model, name in zip(models, model_names):
        score = company_data.get(f'cumulative_ai_score_{model}', 'N/A')
        if score != 'N/A':
            model_scores.append(f"{name}: {score:.1f}")
        else:
            model_scores.append(f"{name}: N/A")
    
    print(f"     💻 Model Scores: {' | '.join(model_scores)}")
    print()
    
    # Show detailed AI factor scores for each model
    ai_factors = [
        'AI Washing Index',
        'Disclosure Sentiment', 
        'Forward-Looking',
        'Risk - External Threats',
        'Risk - Non-Adoption', 
        'Risk - Own Adoption',
        'Strategic Depth',
        'Talent & Investment'
    ]
    
    print(f"     📊 DETAILED AI FACTOR SCORES:")
    print(f"     {'-' * 70}")
    
    for model, model_name in zip(models, model_names):
        print(f"     🤖 {model_name}:")
        
        factor_scores = []
        for factor in ai_factors:
            categorical_col = f"{factor}_{model}"
            numeric_col = f"{factor}_{model}_numeric"
            
            if categorical_col in company_data.index:
                categorical_score = company_data[categorical_col]
                numeric_score = company_data.get(numeric_col, 'N/A')
                
                if pd.notna(categorical_score) and numeric_score != 'N/A':
                    factor_scores.append(f"{factor}: {categorical_score}({numeric_score:.0f})")
                elif pd.notna(categorical_score):
                    factor_scores.append(f"{factor}: {categorical_score}")
                else:
                    factor_scores.append(f"{factor}: N/A")
            else:
                factor_scores.append(f"{factor}: N/A")
        
        # Print factor scores in a formatted way (2 per line for readability)
        for i in range(0, len(factor_scores), 2):
            line_scores = factor_scores[i:i+2]
            if len(line_scores) == 2:
                print(f"          {line_scores[0]:<35} | {line_scores[1]}")
            else:
                print(f"          {line_scores[0]}")
        print()
    
    print()
    
    # Full summaries for all models
    print(f"     📄 FULL AI SUMMARIES:")
    print(f"     {'-' * 50}")
    
    for model, name in zip(models, model_names):
        summary_col = f'Overall Summary_{model}'
        if summary_col in company_data.index and pd.notna(company_data[summary_col]):
            summary = str(company_data[summary_col])
            print(f"     🤖 {name}:")
            # Wrap text nicely
            words = summary.split()
            lines = []
            current_line = []
            current_length = 0
            
            for word in words:
                if current_length + len(word) + 1 <= 100:  # 100 chars per line
                    current_line.append(word)
                    current_length += len(word) + 1
                else:
                    if current_line:
                        lines.append(' '.join(current_line))
                    current_line = [word]
                    current_length = len(word)
            
            if current_line:
                lines.append(' '.join(current_line))
            
            for line in lines:
                print(f"          {line}")
            print()
        else:
            print(f"     🤖 {name}: No summary available")
            print()
    
    # Full key terms for all models
    print(f"     🔑 KEY AI TERMS:")
    print(f"     {'-' * 50}")
    
    for model, name in zip(models, model_names):
        terms_col = f'Key AI Terms_{model}'
        if terms_col in company_data.index and pd.notna(company_data[terms_col]) and str(company_data[terms_col]).strip():
            terms = str(company_data[terms_col])
            print(f"     🏷️  {name}: {terms}")
        else:
            print(f"     🏷️  {name}: No specific AI terms identified")
    
    print(f"\n     {'=' * 80}\n")

def investigate_ai_hypers():
    """
    Loads data and prints details for specific "AI Hyper" companies.
    """
    print("🚀 INVESTIGATING 'AI HYPERS' (HIGH AI SCORE / LOW RETURN)")
    print("=" * 100)
    
    # Load the dataset
    try:
        df = pd.read_csv(file_path)
    except Exception as e:
        print(f"❌ Error loading dataset: {e}")
        return

    # --- Data Preprocessing (same as your original script) ---
    ai_factors = [
        'AI Washing Index', 'Disclosure Sentiment', 'Forward-Looking',
        'Risk - External Threats', 'Risk - Non-Adoption', 'Risk - Own Adoption',
        'Strategic Depth', 'Talent & Investment'
    ]
    models = ['flash1_5', 'flash2_5', 'gpt4o']
    
    for factor in ai_factors:
        for model in models:
            col_name = f"{factor}_{model}"
            if col_name in df.columns:
                numeric_col_name = f"{factor}_{model}_numeric"
                df[numeric_col_name] = df[col_name].apply(convert_letter_grades_to_numeric)
                
    for model in models:
        numeric_cols = [f"{factor}_{model}_numeric" for factor in ai_factors if f"{factor}_{model}_numeric" in df.columns]
        if numeric_cols:
            df[f'cumulative_ai_score_{model}'] = df[numeric_cols].sum(axis=1, skipna=True)

    cumulative_cols = [f'cumulative_ai_score_{model}' for model in models]
    if all(col in df.columns for col in cumulative_cols):
        df['cumulative_ai_score_overall'] = df[cumulative_cols].mean(axis=1, skipna=True)
    
    # Calculate average score (assuming it's the mean of the cumulative score)
    df['average_ai_score_overall'] = df['cumulative_ai_score_overall'] / len(ai_factors)


    # --- Define Target "AI Hyper" Companies ---
    ai_hyper_targets = {
        "Health Care": [
            ('CRVL', 2023),
            ('OPRX', 2021),
            ('OMCL', 2021),
            ('NEO', 2020),
            ('OMCL', 2022)
        ],
        "Industrials": [
            ('TTEC', 2023),
            ('PRLB', 2020),
            ('TTEC', 2022),
            ('DDD', 2022),
            ('SHYF', 2022)
        ],
        "Consumer Discretionary": [
            ('CHGG', 2023),
            ('IRBT', 2022),
            ('REAL', 2021),
            ('SFIX', 2021),
            ('TSCO', 2023)
        ]
    }

    # --- Find and Print Details for Each Target Company ---
    for sector, companies in ai_hyper_targets.items():
        print(f"\n🏢 SECTOR: {sector.upper()}")
        print("=" * 80)
        
        rank = 1
        for ticker, year in companies:
            # Find the specific company row
            company_data = df[(df['Ticker'] == ticker) & (df['Year'] == year)]
            
            if not company_data.empty:
                # Use the first row found (should be unique)
                print_company_details(company_data.iloc[0], rank, sector)
                rank += 1
            else:
                print(f"❌ Could not find data for {ticker} ({year}) in the dataset.")
                print(f"\n     {'=' * 80}\n")


# Execute the investigation
if __name__ == "__main__":
    investigate_ai_hypers()
```

```python
import pandas as pd
import numpy as np
import warnings
warnings.filterwarnings('ignore')

# --- Configuration ---
FILE_PATH = "/Users/daniel/Library/Mobile Documents/com~apple~CloudDocs/Documents/Master Finance/MasterThesis/ThesisData/perfect_balanced_ai_dataset.csv"
MODELS = ['flash1_5', 'flash2_5', 'gpt4o']
MODEL_NAMES = ['Gemini 1.5 Flash', 'Gemini 2.5 Flash', 'GPT-4o']
AI_FACTORS = [
    'AI Washing Index',
    'Disclosure Sentiment',
    'Forward-Looking',
    'Risk - External Threats',
    'Risk - Non-Adoption',
    'Risk - Own Adoption',
    'Strategic Depth',
    'Talent & Investment'
]

def get_top_a_scorers(df, factor):
    """
    Finds the top 3 companies for a given AI factor based on the number of 'A' grades.
    """
    a_score_counts = pd.DataFrame(index=df.index)
    
    # Count how many models gave an 'A' for the current factor
    for model in MODELS:
        col = f"{factor}_{model}"
        if col in df.columns:
            a_score_counts[model] = (df[col].str.strip().str.upper() == 'A')
            
    df['a_score_count'] = a_score_counts.sum(axis=1)

    # --- Selection Logic ---
    # 1. Companies with 3 'A' scores
    top_3 = df[df['a_score_count'] == 3].head(3)
    
    # 2. If not enough, add companies with 2 'A' scores
    if len(top_3) < 3:
        top_2 = df[df['a_score_count'] == 2]
        top_3 = pd.concat([top_3, top_2]).head(3)

    # 3. If still not enough, add companies with 1 'A' score
    if len(top_3) < 3:
        top_1 = df[df['a_score_count'] == 1]
        top_3 = pd.concat([top_3, top_1]).head(3)
        
    return top_3

def print_top_scorers_details(factor, top_companies):
    """
    Prints the detailed information for the top scoring companies for a specific factor.
    """
    print(f"🏆 Top 3 Companies for: '{factor}'")
    print("=" * 80)

    if top_companies.empty:
        print(f"No companies received an 'A' score for the '{factor}' metric.\n")
        return

    rank = 1
    for index, company in top_companies.iterrows():
        print(f"  #{rank}. {company['Company Name']} ({company['Ticker']})")
        print(f"     - Year: {company['Year']}, Filing Date: {company['filingDate']}, Sector: {company['Sector']}")
        
        print("     - Scores, Summaries, and Key Terms ('A' Grades):")
        for model, model_name in zip(MODELS, MODEL_NAMES):
            score_col = f"{factor}_{model}"
            if score_col in company and company[score_col] == 'A':
                summary = company.get(f'Overall Summary_{model}', 'No summary available.')
                key_terms = company.get(f'Key AI Terms_{model}', 'No key terms identified.')

                print(f"       - Model: {model_name}")
                print(f"         - Score: 'A'")
                print(f"         - Summary: {summary}")
                print(f"         - Key Terms: {key_terms}")

        print("-" * 80)
        rank += 1
    print("\n")


def analyze_top_a_scores():
    """
    Main function to load data and run the analysis for each AI factor.
    """
    try:
        df = pd.read_csv(FILE_PATH)
    except FileNotFoundError:
        print(f"Error: The file was not found at {FILE_PATH}")
        return
    except Exception as e:
        print(f"An error occurred while loading the file: {e}")
        return
        
    for factor in AI_FACTORS:
        top_companies = get_top_a_scorers(df.copy(), factor)
        print_top_scorers_details(factor, top_companies)

# --- Execute the Analysis ---
if __name__ == "__main__":
    analyze_top_a_scores()
```

```python
import pandas as pd
import numpy as np
import warnings
warnings.filterwarnings('ignore')

# --- Configuration ---
FILE_PATH = "/Users/daniel/Library/Mobile Documents/com~apple~CloudDocs/Documents/Master Finance/MasterThesis/ThesisData/perfect_balanced_ai_dataset.csv"
MODELS = ['flash1_5', 'flash2_5', 'gpt4o']
MODEL_NAMES = ['Gemini 1.5 Flash', 'Gemini 2.5 Flash', 'GPT-4o']
AI_FACTORS = [
    'AI Washing Index',
    'Disclosure Sentiment',
    'Forward-Looking',
    'Risk - External Threats',
    'Risk - Non-Adoption',
    'Risk - Own Adoption',
    'Strategic Depth',
    'Talent & Investment'
]
TECH_SECTOR_NAME = 'Information Technology'

def get_diverse_top_scorers(df, factor):
    """
    Finds the top 3 companies for a given AI factor, ensuring sector diversity
    and correct final ranking.
    """
    # Calculate 'A' and 'B' score counts for the given factor
    for grade in ['A', 'B']:
        score_counts = pd.DataFrame(index=df.index)
        for model in MODELS:
            col = f"{factor}_{model}"
            if col in df.columns:
                score_counts[model] = (df[col].str.strip().str.upper() == grade)
        df[f'{grade.lower()}_score_count'] = score_counts.sum(axis=1)

    # Separate companies into tech and non-tech groups
    tech_df = df[df['Sector'] == TECH_SECTOR_NAME].copy()
    non_tech_df = df[df['Sector'] != TECH_SECTOR_NAME].copy()
    
    ranking_cols = ['a_score_count', 'b_score_count']

    # Rank both groups internally
    tech_df = tech_df.sort_values(by=ranking_cols, ascending=False)
    non_tech_df = non_tech_df.sort_values(by=ranking_cols, ascending=False)
    
    # Filter out companies with no 'A' or 'B' scores
    tech_df = tech_df[(tech_df['a_score_count'] > 0) | (tech_df['b_score_count'] > 0)]
    non_tech_df = non_tech_df[(non_tech_df['a_score_count'] > 0) | (non_tech_df['b_score_count'] > 0)]

    # --- Selection Logic ---
    # Select the top 1 tech company and top 2 non-tech companies
    top_tech = tech_df.head(1)
    top_non_tech = non_tech_df.head(2)

    # Combine the selections into a single pool
    top_companies_pool = pd.concat([top_non_tech, top_tech])
    
    # --- *** NEW: Re-sort the final pool to ensure correct ranking *** ---
    # This is the key change to fix the ranking order in the final output.
    final_top_list = top_companies_pool.sort_values(by=ranking_cols, ascending=False)
    
    return final_top_list

def print_top_scorers_details_diverse(factor, top_companies):
    """
    Prints the detailed information for the top scoring companies, noting 'A' or 'B' grades.
    """
    print(f"🏆 Top 3 Companies for: '{factor}' (with Sector Diversity)")
    print("=" * 80)

    if top_companies.empty:
        print(f"No companies received an 'A' or 'B' score for the '{factor}' metric.\n")
        return

    rank = 1
    for index, company in top_companies.iterrows():
        print(f"  #{rank}. {company['Company Name']} ({company['Ticker']})")
        print(f"     - Year: {company['Year']}, Filing Date: {company['filingDate']}, Sector: {company['Sector']}")
        
        print("     - Scores, Summaries, and Key Terms ('A' or 'B' Grades):")
        has_score = False
        for model, model_name in zip(MODELS, MODEL_NAMES):
            score_col = f"{factor}_{model}"
            score = company.get(score_col)
            
            if pd.notna(score) and score.strip().upper() in ['A', 'B']:
                has_score = True
                summary = company.get(f'Overall Summary_{model}', 'No summary available.')
                key_terms = company.get(f'Key AI Terms_{model}', 'No key terms identified.')

                print(f"       - Model: {model_name}")
                print(f"         - Score: '{score.strip().upper()}'")
                print(f"         - Summary: {summary}")
                print(f"         - Key Terms: {key_terms}")
        
        if not has_score:
             print("       - No 'A' or 'B' scores found for this entry.")

        print("-" * 80)
        rank += 1
    print("\n")


def analyze_top_scores_with_diversity():
    """
    Main function to load data and run the diversified analysis for each AI factor.
    """
    try:
        df = pd.read_csv(FILE_PATH)
    except FileNotFoundError:
        print(f"Error: The file was not found at {FILE_PATH}")
        return
    except Exception as e:
        print(f"An error occurred while loading the file: {e}")
        return
        
    for factor in AI_FACTORS:
        top_companies = get_diverse_top_scorers(df.copy(), factor)
        print_top_scorers_details_diverse(factor, top_companies)

# --- Execute the Diversified Analysis ---
if __name__ == "__main__":
    analyze_top_scores_with_diversity()
```

```python
import pandas as pd
import numpy as np
import warnings
warnings.filterwarnings('ignore')

# --- Configuration ---
FILE_PATH = "/Users/daniel/Library/Mobile Documents/com~apple~CloudDocs/Documents/Master Finance/MasterThesis/ThesisData/perfect_balanced_ai_dataset.csv"
MODELS = ['flash1_5', 'flash2_5', 'gpt4o']
MODEL_NAMES = ['Gemini 1.5 Flash', 'Gemini 2.5 Flash', 'GPT-4o']
TECH_SECTOR_NAME = 'Information Technology'

# --- CORRECTED: The list of factors with the exact names from your file ---
AI_FACTORS = [
    'AI Washing Index',
    'Disclosure Sentiment',
    'Forward-Looking',
    'Risk - External Threats',
    'Risk - Non-Adoption',
    'Risk - Own Adoption',
    'Strategic Depth',  # <-- The corrected name
    'Talent & Investment'
]

# --- CORRECTED: Maps now use the correct key 'Strategic Depth' ---
SECOND_TO_LAST_GRADE_MAP = {
    'AI Washing Index': 'D',
    'Disclosure Sentiment': 'D',
    'Forward-Looking': 'C',
    'Risk - Own Adoption': 'B',
    'Risk - External Threats': 'B',
    'Risk - Non-Adoption': 'B',
    'Strategic Depth': 'D', # <-- Corrected key
    'Talent & Investment': 'C',
}

LAST_GRADE_MAP = {
    'AI Washing Index': 'E',
    'Disclosure Sentiment': 'E',
    'Forward-Looking': 'D',
    'Risk - Own Adoption': 'C',
    'Risk - External Threats': 'C',
    'Risk - Non-Adoption': 'C',
    'Strategic Depth': 'E', # <-- Corrected key
    'Talent & Investment': 'D',
}


def get_low_scorers_diverse(df, factor, target_grade):
    """
    Finds the top 3 companies with the most instances of a specific target grade
    for a given AI factor, while ensuring sector diversity and correct final ranking.
    """
    count_col = 'target_grade_count'
    score_counts = pd.DataFrame(index=df.index)
    for model in MODELS:
        col = f"{factor}_{model}"
        if col in df.columns:
            score_counts[model] = (df[col].str.strip().str.upper() == target_grade)
    df[count_col] = score_counts.sum(axis=1)

    tech_df = df[df['Sector'] == TECH_SECTOR_NAME].copy()
    non_tech_df = df[df['Sector'] != TECH_SECTOR_NAME].copy()

    tech_df = tech_df.sort_values(by=[count_col], ascending=False)
    non_tech_df = non_tech_df.sort_values(by=[count_col], ascending=False)

    tech_df = tech_df[tech_df[count_col] > 0]
    non_tech_df = non_tech_df[non_tech_df[count_col] > 0]

    top_tech = tech_df.head(1)
    top_non_tech = non_tech_df.head(2)

    top_companies_pool = pd.concat([top_non_tech, top_tech])
    final_top_list = top_companies_pool.sort_values(by=[count_col], ascending=False)

    return final_top_list

def print_low_scorers_details(factor, top_companies, target_grade):
    """
    Prints the detailed information for the top low-scoring companies.
    """
    print(f"🏆 Top Companies with the Most '{target_grade}' Scores for: '{factor}' (with Sector Diversity)")
    print("=" * 80)

    if top_companies.empty:
        print(f"No companies received a '{target_grade}' score for the '{factor}' metric.\n")
        return

    rank = 1
    for index, company in top_companies.iterrows():
        print(f"  #{rank}. {company['Company Name']} ({company['Ticker']})")
        print(f"     - Year: {company['Year']}, Filing Date: {company['filingDate']}, Sector: {company['Sector']}")
        print(f"     - Scores, Summaries, and Key Terms (Showing '{target_grade}' Grades):")

        for model, model_name in zip(MODELS, MODEL_NAMES):
            score_col = f"{factor}_{model}"
            score = company.get(score_col)

            if pd.notna(score) and score.strip().upper() == target_grade:
                summary = company.get(f'Overall Summary_{model}', 'No summary available.')
                key_terms = company.get(f'Key AI Terms_{model}', 'No key terms identified.')
                print(f"       - Model: {model_name}")
                print(f"         - Score: '{target_grade}'")
                print(f"         - Summary: {summary}")
                print(f"         - Key Terms: {key_terms}")
        print("-" * 80)
        rank += 1
    print("\n")


def analyze_low_scores_with_diversity():
    """
    Main function to load data and run the analysis for each AI factor.
    Includes fallback logic if no companies are found for the primary target grade.
    """
    try:
        df = pd.read_csv(FILE_PATH)
    except FileNotFoundError:
        print(f"Error: The file was not found at {FILE_PATH}")
        return
    except Exception as e:
        print(f"An error occurred while loading the file: {e}")
        return

    for factor in AI_FACTORS:
        primary_target_grade = SECOND_TO_LAST_GRADE_MAP.get(factor)
        if not primary_target_grade:
            print(f"Warning: No target grade defined for factor '{factor}'. Skipping.")
            continue

        top_companies = get_low_scorers_diverse(df.copy(), factor, primary_target_grade)

        if top_companies.empty:
            fallback_grade = LAST_GRADE_MAP.get(factor)
            print(f"INFO: No companies found with grade '{primary_target_grade}' for '{factor}'.")
            print(f"      ---> Now searching for last-place grade '{fallback_grade}' instead.\n")

            if fallback_grade:
                top_companies = get_low_scorers_diverse(df.copy(), factor, fallback_grade)
                print_low_scorers_details(factor, top_companies, fallback_grade)
            else:
                 print(f"Warning: No fallback grade defined for factor '{factor}'. No results to show.")
        else:
            print_low_scorers_details(factor, top_companies, primary_target_grade)


# --- Execute the Analysis ---
if __name__ == "__main__":
    analyze_low_scores_with_diversity()
```

```python
import pandas as pd
import numpy as np

def find_interesting_cases(reg_data, sector_name='Consumer Discretionary', return_col='excess_return_12mo'):
    """
    Identifies and analyzes firms within a specific sector that have high AI scores
    but subsequently underperform, providing concrete examples for case studies.
    """
    print("🚀 CASE STUDY EXPLORER")
    print("=" * 50)
    print(f"🎯 Objective: Find firms in '{sector_name}' with HIGH AI scores and LOW subsequent returns.")
    
    # 1. Filter the dataset to the specified sector
    sector_df = reg_data[reg_data['Sector'] == sector_name].copy()
    
    if sector_df.empty:
        print(f"\n❌ Error: No data found for the '{sector_name}' sector. Please check the sector name.")
        return None
        
    print(f"\n1. Filtering to '{sector_name}': Found {len(sector_df):,} observations.")
    
    # 2. Define the thresholds for "high" and "low" using quartiles
    # We use dropna() to ensure quantiles are calculated on valid data
    high_ai_threshold = sector_df['AI_Cumulative_Score'].dropna().quantile(0.75)
    low_return_threshold = sector_df[return_col].dropna().quantile(0.25)
    
    print(f"2. Defining Thresholds:")
    print(f"   - 'High AI Score' is defined as > {high_ai_threshold:.2f} (Top 25%)")
    print(f"   - 'Low Return' is defined as < {low_return_threshold:.2%} (Bottom 25%)")
    
    # 3. Identify the firms that meet both criteria
    interesting_cases = sector_df[
        (sector_df['AI_Cumulative_Score'] > high_ai_threshold) &
        (sector_df[return_col] < low_return_threshold)
    ]
    
    if interesting_cases.empty:
        print("\n✅ No specific firms were found that met both the 'High AI' and 'Low Return' criteria.")
        return None
        
    print(f"\n3. Identifying Cases: Found {len(interesting_cases)} observations meeting both criteria.")
    
    # 4. Prepare a clean table for analysis
    # Select key columns for the report
    display_cols = [
        'Ticker', 'Company Name', 'Year', 'AI_Cumulative_Score', return_col
    ]
    
    # Also find out which individual AI factor was highest for each case
    ai_factor_cols = [
        'AI_Washing', 'Disclosure_Sentiment', 'Forward_Looking',
        'Strategic_Depth', 'Talent_Investment', 'Risk_External',
        'Risk_Non_Adoption', 'Risk_Own_Adoption'
    ]
    # Ensure we only use factors present in the data
    available_ai_factors = [f for f in ai_factor_cols if f in interesting_cases.columns]
    
    if available_ai_factors:
        interesting_cases['Top_AI_Factor'] = interesting_cases[available_ai_factors].idxmax(axis=1)
        display_cols.append('Top_AI_Factor')

    report_table = interesting_cases[display_cols].sort_values(by=[return_col], ascending=True)
    
    # Format the return column as a percentage
    report_table[return_col] = report_table[return_col].apply(lambda x: f"{x:.2%}")

    print("\n\n📊 TABLE OF INTERESTING CASES")
    print("---------------------------------------------------------------------------------------------------")
    print("The following companies had AI scores in the top quartile but returns in the bottom quartile.")
    print("These are excellent candidates for a qualitative case study in your thesis.")
    print("---------------------------------------------------------------------------------------------------")
    print(report_table.to_string(index=False))
    
    return report_table

#================================================================================
#                                 RUN THE ANALYSIS
#================================================================================
# This calls the function using your 'final_reg_data' dataframe.

# Make sure 'final_reg_data' is loaded and available in your notebook
if 'final_reg_data' in locals():
    interesting_cases_df = find_interesting_cases(final_reg_data)
else:
    print("\n\n❌ ERROR: The 'final_reg_data' dataframe was not found.")
    print("Please ensure you have run the preceding cells in your notebook to create it.")

```

```python
import pandas as pd
import numpy as np

def run_sector_quadrant_analysis(reg_data, return_col='excess_return_12mo', quantile_threshold=0.25):
    """
    Performs a comprehensive quadrant analysis for every sector to find outliers based on
    AI scores and subsequent returns.

    Args:
        reg_data (pd.DataFrame): The final regression dataframe.
        return_col (str): The return column to use for the analysis.
        quantile_threshold (float): The percentage to use for defining high/low quantiles (e.g., 0.25 for quartiles).
    """
    print("🚀 SECTOR OUTLIER & QUADRANT ANALYSIS")
    print("=" * 60)
    print(f"🎯 Objective: For each sector, find firms in all four quadrants of AI Score vs. Return.")
    print(f"   (Using top/bottom {int(quantile_threshold*100)}% for high/low thresholds)")

    all_sectors = sorted([s for s in reg_data['Sector'].unique() if pd.notna(s)])
    all_outliers_list = []

    for sector in all_sectors:
        print(f"\n\nANALYZING SECTOR: {sector.upper()}")
        print("-" * 50)
        
        sector_df = reg_data[reg_data['Sector'] == sector].copy()
        if len(sector_df) < 20:
            print("Skipping due to insufficient observations.")
            continue

        # Define high/low thresholds for AI Score and Returns for the current sector
        high_ai_thresh = sector_df['AI_Cumulative_Score'].quantile(1 - quantile_threshold)
        low_ai_thresh = sector_df['AI_Cumulative_Score'].quantile(quantile_threshold)
        high_ret_thresh = sector_df[return_col].quantile(1 - quantile_threshold)
        low_ret_thresh = sector_df[return_col].quantile(quantile_threshold)

        # Identify firms in each of the four quadrants
        quadrants = {
            "High AI / High Return (Champions)": sector_df[(sector_df['AI_Cumulative_Score'] >= high_ai_thresh) & (sector_df[return_col] >= high_ret_thresh)],
            "High AI / Low Return (AI Hypers)": sector_df[(sector_df['AI_Cumulative_Score'] >= high_ai_thresh) & (sector_df[return_col] <= low_ret_thresh)],
            "Low AI / High Return (Quiet Performers)": sector_df[(sector_df['AI_Cumulative_Score'] <= low_ai_thresh) & (sector_df[return_col] >= high_ret_thresh)],
            "Low AI / Low Return (Laggards)": sector_df[(sector_df['AI_Cumulative_Score'] <= low_ai_thresh) & (sector_df[return_col] <= low_ret_thresh)],
        }
        
        # Display results and collect data for export
        for quad_name, quad_df in quadrants.items():
            print(f"\n✅ {quad_name}: Found {len(quad_df)} firms.")
            if not quad_df.empty:
                # Prepare table for display
                display_cols = ['Ticker', 'Company Name', 'Year', 'AI_Cumulative_Score', return_col]
                report_table = quad_df[display_cols].copy()
                # Sort by return to see the most extreme examples
                sort_asc = True if 'Low Return' in quad_name else False
                report_table = report_table.sort_values(by=return_col, ascending=sort_asc)
                report_table[return_col] = report_table[return_col].apply(lambda x: f"{x:.2%}")
                
                print(report_table.head().to_string(index=False))
                if len(report_table) > 5:
                    print(f"... and {len(report_table)-5} more.")

                # Add data to our master list for final export
                for _, row in quad_df.iterrows():
                    all_outliers_list.append({
                        'Sector': sector,
                        'Quadrant': quad_name,
                        'Ticker': row['Ticker'],
                        'Company Name': row['Company Name'],
                        'Year': row['Year'],
                        'AI_Cumulative_Score': row['AI_Cumulative_Score'],
                        'Return': row[return_col]
                    })
    
    # Export the complete list of outliers to a single CSV file
    if all_outliers_list:
        outliers_df = pd.DataFrame(all_outliers_list)
        filename = "sector_quadrant_analysis_all_outliers.csv"
        outliers_df.to_csv(filename, index=False)
        print(f"\n\n\n✅✅✅ Complete list of all outliers from all sectors exported to: {filename} ✅✅✅")

#================================================================================
#                                 RUN THE ANALYSIS
#================================================================================
# This calls the explorer function using your 'final_reg_data' dataframe.

if 'final_reg_data' in locals():
    # You can change the quantile_threshold to find more or fewer outliers (e.g., 0.2 for top/bottom 20%)
    run_sector_quadrant_analysis(final_reg_data, quantile_threshold=0.25)
else:
    print("\n\n❌ ERROR: The 'final_reg_data' dataframe was not found.")
    print("Please ensure you have run the preceding cells in your notebook to create it.")
```
